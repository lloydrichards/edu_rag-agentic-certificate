# Prompt Engineering: Zero-shot, One-shot, and Few-shot

Experiments with different prompting techniques to understand their impact on model performance.

**Requires**: `OPENAI_API_KEY` in your environment or `.env`.

```python
from langchain_openai import ChatOpenAI
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate

def llm_model(prompt_txt, params=None):
    
    default_params = {
        "temperature": 0.5,
    }
    
    prompt = ChatPromptTemplate.from_template("{prompt_txt}")
    
    model = ChatOpenAI(model="gpt-4o-mini", **(params or default_params))
    
    parser = StrOutputParser()
    
    chain = prompt | model | parser
    
    return chain.invoke(prompt_txt)
```

## Zero-shot Prompting

```python
prompt = """Answer the following question:
            What is the capital of France?
            
            Answer:"""

response = llm_model(prompt)
print(response)
```

## One-shot Prompting

```python
prompt = """Here is an example of a good movie and food pairing:
            Movie: The Godfather
            Food: Spaghetti and Meatballs

            Now, provide a good movie and food pairing for the following movie:
            Movie: The Shawshank Redemption
            Answer:"""

response = llm_model(prompt)
print(response)
```

## Chain-of-Thought Prompting

```python
prompt = """Explain how to make a PB&J sandwich step-by-step. 
            For each step, briefly say why it is necessary so I don't make a mess.
            """

response = llm_model(prompt)
print(response)
```
