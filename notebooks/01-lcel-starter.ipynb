# LangChain LCEL Chaining Method

Minimal example that composes a prompt, model, and output parser using LCEL.

**Requires**: `OPENAI_API_KEY` in your environment or `.env`.

```python
from langchain_openai import ChatOpenAI
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate

model = ChatOpenAI(model="gpt-4o-mini")

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a concise tutor."),
    ("human", "Explain {topic} in 3 bullet points.")
])

parser = StrOutputParser()

chain = prompt | model | parser
#              ^? pipe operators are used to connect the components of the chain
```

```python
chain.invoke({"topic": "retrieval-augmented generation (RAG)"})
```

## Optional: add simple post-processing
LCEL makes it easy to add steps. Here we strip whitespace.

```python
chain = prompt | model | parser | (lambda x: x.strip())
chain.invoke({"topic": "vector databases"})
```
